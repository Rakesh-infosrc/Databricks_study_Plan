## 1. What is Lazy Evaluation?

Spark does **not** execute operations immediately.

Instead, it **records** all your steps and waits.
Only when you call an **Action** does Spark execute everything **together** in one optimized run.

> Think of it like a waiter writing down your full order before going to the kitchen — not running after every item.

## 2. Transformations

Transformations **build the plan** — they do NOT run immediately.

| Transformation | Purpose |
|---|---|
| `filter()` / `where()` | Filter rows |
| `select()` | Pick columns |
| `withColumn()` | Add/modify column |
| `groupBy()` | Group records |
| `orderBy()` | Sort data |
| `join()` | Join DataFrames |
| `distinct()` | Remove duplicates |

```python
# TRANSFORMATIONS — Spark does NOT execute these yet

data = [
    ("John", 25, 50000),
    ("Alice", 32, 70000),
    ("David", 40, 90000),
    ("Emma", 28, 60000)
]

columns = ["Name", "Age", "Salary"]

df = spark.createDataFrame(data, columns)

# Step 1: Filter — no execution yet
filtered_df = df.filter("Age > 30")

# Step 2: Select — no execution yet
selected_df = filtered_df.select("Name", "Salary")

# Step 3: Sort — no execution yet
result = selected_df.orderBy("Salary")

print(" Transformations recorded. Execution has NOT started yet.")
```

## 3. Actions

Actions **trigger execution** of the full plan Spark built.

| Action | Purpose |
|---|---|
| `show()` | Print rows to console |
| `display()` | Databricks rich table view |
| `count()` | Count total rows |
| `collect()` | Return all rows as Python list |
| `first()` | Return first row |
| `write()` | Save output to storage |

```python
#  ACTION — This triggers Spark to execute everything!
# Spark now runs: Read → Filter → Select → Sort → Display

print(" Action triggered — Spark begins execution...")
display(result)
```

## 4. Step-by-Step Walkthrough

Watch how Spark builds the plan one step at a time, then executes only at the end.

```python
# STEP 1
df = spark.createDataFrame(data, columns)
print("Step 1  DataFrame defined — no execution yet")

# STEP 2
df2 = df.filter("Age > 30")
print("Step 2  Filter recorded — no execution yet")

# STEP 3
df3 = df2.select("Name", "Salary")
print("Step 3  Select recorded — no execution yet")

# STEP 4 — Action!
print("Step 4  show() called — Spark executes everything now!")
df3.show()
```

## 5. View the Execution Plan

Use `explain()` to see exactly what plan Spark built before executing.

- **Parsed Plan** → your raw code as Spark understood it
- **Analyzed Plan** → after resolving column names and types
- **Optimized Plan** → after Catalyst Optimizer improves it
- **Physical Plan** → the actual execution strategy

```python
# View the Logical + Physical plan Spark builds
result.explain(extended=True)
```

## ✅ Key Takeaways

| Concept | Description |
|---|---|
| **Transformation** | Builds the plan — no execution |
| **Action** | Triggers full execution |
| **Lazy Evaluation** | Spark waits for an Action before running |
| **Catalyst Optimizer** | Optimizes the plan before execution |
| **Logical Plan** | What you asked Spark to do |
| **Physical Plan** | How Spark will actually do it |

> **Rule:** Transformations build the plan. Actions execute the plan.
