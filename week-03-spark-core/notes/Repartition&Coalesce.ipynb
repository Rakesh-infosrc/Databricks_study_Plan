## 1. What is Repartition?

`repartition()` changes the number of partitions by **redistributing data** across the cluster.

- Can **increase or decrease** partitions
- Performs a **full shuffle** — expensive
- Use when you need to increase partitions or evenly balance data

```python
# Repartition Example

df = spark.range(100)
print("Before:", df.rdd.getNumPartitions())

df = df.repartition(8)
print("After repartition(8):", df.rdd.getNumPartitions())
```

## 2. What is Coalesce?

`coalesce()` reduces partitions by **merging existing ones**.

- Can only **decrease** partitions
- Usually avoids a full shuffle — faster
- Use when reducing partitions after a filter or before writing output

```python
# Coalesce Example

df = spark.range(100)
print("Before:", df.rdd.getNumPartitions())

df = df.coalesce(2)
print("After coalesce(2):", df.rdd.getNumPartitions())
```

## 3. Quick Comparison

| Feature | Repartition | Coalesce |
|---|---|---|
| Increase Partitions | ✅ | ❌ |
| Decrease Partitions | ✅ | ✅ |
| Shuffle | Yes — full shuffle | Usually No |
| Performance | Slower | Faster |
| Use Case | Rebalance or increase | Reduce before write |

```python
# Side-by-side comparison

df = spark.range(100)
print("Original partitions:", df.rdd.getNumPartitions())

df_repartition = df.repartition(8)
print("After repartition(8):", df_repartition.rdd.getNumPartitions())

df_coalesce = df.coalesce(2)
print("After coalesce(2):", df_coalesce.rdd.getNumPartitions())
```

## ✅ Key Takeaways

| Concept | Description |
|---|---|
| **repartition()** | Increases or decreases partitions — always shuffles |
| **coalesce()** | Decreases partitions only — avoids shuffle |
| **Performance** | `coalesce()` is faster for reducing partitions |
| **Best Practice** | Use `coalesce()` before writing output to reduce files |

> **Interview Answer:** `coalesce()` is generally faster because it avoids a full shuffle by merging existing partitions instead of redistributing all data.
